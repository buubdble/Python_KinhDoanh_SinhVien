### A2. Tính giá bán sau chiết khấu và thuế VAT
**Tình huống:** Cửa hàng bán một chiếc tai nghe giá niêm yết 850.000 đ, giảm 15%, sau đó cộng VAT 8%. Khách mua 3 chiếc.

**Bước 1 – Khai báo biến.** Tên biến viết thường, nối bằng dấu `_`, đặt tên có nghĩa.

```python
gia_niem_yet = 850_000      # đồng (dấu _ giúp dễ đọc số lớn)
ty_le_giam = 0.15           # 15%
vat = 0.08                  # 8%
so_luong = 3
print(type(gia_niem_yet), type(ty_le_giam), type(so_luong))
```

**Bước 2 – Tính toán** với các toán tử `+ - * /`.

```python
gia_sau_giam = gia_niem_yet * (1 - ty_le_giam)
gia_sau_vat = gia_sau_giam * (1 + vat)
thanh_tien = gia_sau_vat * so_luong
print(gia_sau_giam, gia_sau_vat, thanh_tien)
```

**Bước 3 – Định dạng kết quả.** f-string `f"...{bien:,.0f}..."` chèn giá trị vào chuỗi;
`:,.0f` = dấu phẩy ngăn cách hàng nghìn, 0 chữ số thập phân. Thay `,` bằng `.` để đúng kiểu Việt Nam.

```python
def vnd(x):
    return f"{x:,.0f}".replace(",", ".") + " đ"

print("Giá sau giảm :", vnd(gia_sau_giam))
print("Giá sau VAT  :", vnd(gia_sau_vat))
print("Thành tiền   :", vnd(thanh_tien))
```

### A3. Phép chia lấy nguyên `//`, chia lấy dư `%` và lũy thừa `**`
**Tình huống:** Kho có 1.250 chai nước, mỗi thùng đóng 24 chai. Được bao nhiêu thùng, dư bao nhiêu chai?

```python
tong_chai = 1250
chai_moi_thung = 24
so_thung = tong_chai // chai_moi_thung
du = tong_chai % chai_moi_thung
print(f"Đóng được {so_thung} thùng, còn dư {du} chai")
```

**Lãi kép:** Gửi 100 triệu đồng, lãi suất 6%/năm, sau 5 năm được bao nhiêu? Công thức: `A = P × (1 + r)^n`.

```python
P, r, n = 100_000_000, 0.06, 5
A = P * (1 + r) ** n
print("Số tiền sau", n, "năm:", vnd(A))
print("Tiền lãi:", vnd(A - P))
```

### Bài B1. Hóa đơn bán hàng

Khách mua 2 áo sơ mi (350.000 đ/chiếc) và 1 quần tây (520.000 đ). Cửa hàng giảm 10% trên tổng tiền hàng,
sau đó tính VAT 8%. In ra: tổng tiền hàng, tiền giảm, tiền VAT, số tiền khách phải trả (định dạng `1.234.567 đ`).

```python
gia_ao = 350000
so_ao = 2
gia_quan = 520000

tong_tien = gia_ao * so_ao + gia_quan
tien_giam = tong_tien * 0.10
tien_sau_giam = tong_tien - tien_giam
tien_vat = tien_sau_giam * 0.08
thanh_toan = tien_sau_giam + tien_vat

print("Tổng tiền hàng:", f"{tong_tien:,.0f}".replace(",", "."), "đ")
print("Tiền giảm:", f"{tien_giam:,.0f}".replace(",", "."), "đ")
print("Tiền VAT:", f"{tien_vat:,.0f}".replace(",", "."), "đ")
print("Khách phải trả:", f"{thanh_toan:,.0f}".replace(",", "."), "đ")
```

### Bài B2. Điểm hòa vốn

Một quán cà phê có định phí 45 triệu đồng/tháng, giá bán trung bình 35.000 đ/ly, biến phí 14.000 đ/ly.
Tính số ly cần bán mỗi tháng để hòa vốn (làm tròn lên) và số ly mỗi ngày (tháng 30 ngày).
Công thức: `Q = Định phí / (Giá bán − Biến phí)`.

<details><summary>💡 Gợi ý</summary>

Dùng `math.ceil()` để làm tròn lên (cần `import math`).

</details>

```python
import math

dinh_phi = 45000000
gia_ban = 35000
bien_phi = 14000

so_ly_thang = math.ceil(dinh_phi / (gia_ban - bien_phi))
so_ly_ngay = math.ceil(so_ly_thang / 30)

print("Số ly mỗi tháng:", so_ly_thang)
print("Số ly mỗi ngày:", so_ly_ngay)
```

### Bài B3. Trả góp hàng tháng

Khách mua laptop 24.000.000 đ trả góp 12 tháng, lãi suất 1,2%/tháng. Tính số tiền trả mỗi tháng theo công thức
niên kim: `PMT = P × r × (1 + r)^n / ((1 + r)^n − 1)`, tổng số tiền trả và tổng tiền lãi.

```python
P = 24000000
r = 0.012
n = 12

tien_moi_thang = P * r * (1 + r)**n / ((1 + r)**n - 1)
tong_tien = tien_moi_thang * n
tong_lai = tong_tien - P

print(f"Trả mỗi tháng: {tien_moi_thang:,.0f} đ")
print(f"Tổng tiền trả: {tong_tien:,.0f} đ")
print(f"Tổng tiền lãi: {tong_lai:,.0f} đ")
```

### Bài B4. Chuẩn hóa mã sản phẩm

Cho chuỗi `ma = "  sp-dt-00045 "`. Hãy: (a) bỏ khoảng trắng và viết hoa → `SP-DT-00045`;
(b) tách lấy mã nhóm hàng `DT`; (c) lấy số thứ tự dưới dạng số nguyên `45`;
(d) tạo mã mới cho sản phẩm kế tiếp `SP-DT-00046` (giữ đủ 5 chữ số).

<details><summary>💡 Gợi ý</summary>

`f"{so:05d}"` in số nguyên đủ 5 chữ số, thêm số 0 phía trước.

</details>

```python
ma = "  sp-dt-00045 "

ma_chuan = ma.strip().upper()
cac_phan = ma_chuan.split("-")

nhom_hang = cac_phan[1]
so_thu_tu = int(cac_phan[2])

ma_moi = f"SP-{nhom_hang}-{so_thu_tu + 1:05d}"

print("Mã chuẩn:", ma_chuan)
print("Nhóm hàng:", nhom_hang)
print("Số thứ tự:", so_thu_tu)
print("Mã mới:", ma_moi)
```

### Bài B5. Tăng trưởng doanh thu

Doanh thu quý 1 là 1,85 tỷ đồng, quý 2 là 2,12 tỷ đồng. Tính tỷ lệ tăng trưởng (%) và in dạng `Tăng trưởng: 14.6%`.
Nếu duy trì tốc độ này, doanh thu quý 4 dự kiến bao nhiêu?

```python
quy_1 = 1.85
quy_2 = 2.12

tang_truong = (quy_2 - quy_1) / quy_1

quy_3 = quy_2 * (1 + tang_truong)
quy_4 = quy_3 * (1 + tang_truong)

print(f"Tăng trưởng: {tang_truong:.1%}")
print(f"Doanh thu quý 4: {quy_4:.3f} tỷ đồng")
```

### Bài C2. Đọc thông tin từ chuỗi đơn hàng

Cho chuỗi `don = "DH2025-1123|Nguyễn Thị Hoa|0903123456|3|189000"` (mã đơn | tên | SĐT | số lượng | đơn giá).
Tách các trường, che số điện thoại thành `0903***456` và in hóa đơn gọn gàng gồm thành tiền.

```python
don = "DH2025-1123|Nguyễn Thị Hoa|0903123456|3|189000"

thong_tin = don.split("|")

ma_don = thong_tin[0]
ten = thong_tin[1]
sdt = thong_tin[2]
so_luong = int(thong_tin[3])
don_gia = int(thong_tin[4])

sdt_che = sdt[:4] + "***" + sdt[-3:]
thanh_tien = so_luong * don_gia

print("HÓA ĐƠN BÁN HÀNG")
print("Mã đơn:", ma_don)
print("Khách hàng:", ten)
print("SĐT:", sdt_che)
print("Số lượng:", so_luong)
print("Đơn giá:", f"{don_gia:,.0f}", "đồng")
print("Thành tiền:", f"{thanh_tien:,.0f}", "đồng")
```

### Bài C3. Quy tắc 72

Quy tắc 72 ước lượng số năm để vốn tăng gấp đôi: `72 / lãi suất (%)`. Với lãi suất 4%, 6%, 8%, 12%,
hãy so sánh kết quả ước lượng với kết quả chính xác `ln(2) / ln(1 + r)`. (Có thể viết lặp lại 4 lần;
buổi sau ta sẽ dùng vòng lặp.)

```python
import math

# Lãi suất 4%
uoc_luong_4 = 72 / 4
chinh_xac_4 = math.log(2) / math.log(1 + 0.04)

# Lãi suất 6%
uoc_luong_6 = 72 / 6
chinh_xac_6 = math.log(2) / math.log(1 + 0.06)

# Lãi suất 8%
uoc_luong_8 = 72 / 8
chinh_xac_8 = math.log(2) / math.log(1 + 0.08)

# Lãi suất 12%
uoc_luong_12 = 72 / 12
chinh_xac_12 = math.log(2) / math.log(1 + 0.12)

print("Lãi suất 4%:", uoc_luong_4, "và", round(chinh_xac_4, 2))
print("Lãi suất 6%:", uoc_luong_6, "và", round(chinh_xac_6, 2))
print("Lãi suất 8%:", uoc_luong_8, "và", round(chinh_xac_8, 2))
print("Lãi suất 12%:", uoc_luong_12, "và", round(chinh_xac_12, 2))
```
