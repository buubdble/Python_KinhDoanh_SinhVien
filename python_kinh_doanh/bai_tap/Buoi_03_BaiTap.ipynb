### Bài B3. Doanh thu theo nhóm hàng

Tính tổng doanh thu, số hóa đơn và tỷ trọng doanh thu (%) của từng nhóm hàng, sắp xếp giảm dần.
Nhóm nào đóng góp nhiều nhất?

<details><summary>💡 Gợi ý</summary>

`df.groupby('nhom_hang')['tong_tien'].sum()` cho tổng; chia cho tổng chung để ra tỷ trọng.

</details>

```python
# Viết code của bạn ở đây
thong_ke = df.groupby("nhom_hang").agg(
    tong_doanh_thu=("tong_tien", "sum"),
    so_hoa_don=("ma_hoa_don", "count")
)

thong_ke["ty_trong"] = (
    thong_ke["tong_doanh_thu"] / df["tong_tien"].sum() * 100
).round(2)

thong_ke = thong_ke.sort_values("tong_doanh_thu", ascending=False)

print(thong_ke)
print("\nNhóm có doanh thu cao nhất:", thong_ke.index[0])
```

### Bài B4. Cột tính toán mới

Tạo cột `muc_danh_gia` nhận giá trị "Tốt" nếu `danh_gia` ≥ 8, "Trung bình" nếu từ 6 đến dưới 8, "Kém" nếu < 6.
Đếm số hóa đơn theo từng mức và tính tỷ lệ phần trăm.

<details><summary>💡 Gợi ý</summary>

Dùng `np.where(dieu_kien, gia_tri_dung, gia_tri_sai)` lồng nhau, hoặc `pd.cut`.

</details>

```python
# Viết code của bạn ở đây

import numpy as np

df["muc_danh_gia"] = np.where(
    df["danh_gia"] >= 8, "Tốt",
    np.where(df["danh_gia"] >= 6, "Trung bình", "Kém")
)

so_luong = df["muc_danh_gia"].value_counts()
ty_le = (so_luong / len(df) * 100).round(2)

ket_qua = pd.DataFrame({
    "Số hóa đơn": so_luong,
    "Tỷ lệ (%)": ty_le
})

print(ket_qua)
```

### Bài C1. Khung giờ vàng

Từ cột `gio` (dạng "HH:MM"), tạo cột `khung_gio` (lấy giờ dạng số nguyên). Khung giờ nào có tổng doanh thu cao nhất?
Chia ngày thành Sáng (8–11), Trưa (12–14), Chiều (15–17), Tối (18–21) và so sánh doanh thu trung bình mỗi hóa đơn.

```python
# Viết code của bạn ở đây

# Lấy giờ từ cột gio
df["khung_gio"] = df["gio"].str[:2].astype(int)

# Tính doanh thu theo từng giờ
doanh_thu_gio = df.groupby("khung_gio")["tong_tien"].sum()

print("Giờ có doanh thu cao nhất:", doanh_thu_gio.idxmax())
print("Doanh thu:", round(doanh_thu_gio.max(), 2), "nghìn đồng")

# Chia các khung giờ thành từng buổi
def chia_buoi(gio):
    if 8 <= gio <= 11:
        return "Sáng"
    elif 12 <= gio <= 14:
        return "Trưa"
    elif 15 <= gio <= 17:
        return "Chiều"
    elif 18 <= gio <= 21:
        return "Tối"
    else:
        return "Khác"

df["buoi"] = df["khung_gio"].apply(chia_buoi)

# Tính giá trị hóa đơn trung bình theo buổi
ket_qua = df[df["buoi"] != "Khác"].groupby("buoi")["tong_tien"].mean()

print("\nGiá trị hóa đơn trung bình theo buổi:")
print(ket_qua.round(2))
```

### Bài C2. NumPy: mô phỏng giá cổ phiếu

Mô phỏng giá một cổ phiếu trong 250 phiên: giá ban đầu 50.000 đ, lợi suất mỗi phiên ~ phân phối chuẩn
trung bình 0,05%, độ lệch chuẩn 2% (`np.random.default_rng(0).normal`). Dùng `np.cumprod` để tính giá,
cho biết giá cuối, giá cao nhất, mức sụt giảm lớn nhất từ đỉnh (max drawdown).

```python
# Viết code của bạn ở đây

import numpy as np

rng = np.random.default_rng(0)

gia_ban_dau = 50000

# Tạo lợi suất ngẫu nhiên cho 250 phiên
loi_suat = rng.normal(0.0005, 0.02, 250)

# Tính giá cổ phiếu qua từng phiên
gia = gia_ban_dau * np.cumprod(1 + loi_suat)

# Tìm giá cao nhất tính đến từng phiên
gia_cao_nhat = np.maximum.accumulate(gia)

# Tính mức sụt giảm từ đỉnh
sut_giam = (gia - gia_cao_nhat) / gia_cao_nhat * 100

print("Giá cuối kỳ: %.0f đồng" % gia[-1])
print("Giá cao nhất: %.0f đồng" % gia.max())
print("Mức sụt giảm lớn nhất: %.2f%%" % sut_giam.min())
```

### Bài C3. Tự viết hàm báo cáo

Viết hàm `bao_cao(df, cot_nhom)` trả về DataFrame gồm: số hóa đơn, tổng doanh thu, doanh thu TB/hóa đơn,
tổng lợi nhuận gộp, biên lợi nhuận gộp (%) theo cột nhóm bất kỳ. Gọi hàm với `thanh_pho`, `thanh_toan`, `loai_khach`.

```python
# Viết code của bạn ở đây

def bao_cao(df, cot_nhom):
    kq = df.groupby(cot_nhom).agg(
        so_hoa_don=("ma_hoa_don", "count"),
        tong_doanh_thu=("tong_tien", "sum"),
        doanh_thu_tb=("tong_tien", "mean"),
        tong_loi_nhuan=("loi_nhuan_gop", "sum")
    )

    kq["bien_loi_nhuan (%)"] = (
        kq["tong_loi_nhuan"] / kq["tong_doanh_thu"] * 100
    ).round(2)

    return kq.round(2)

print("Báo cáo theo thành phố:")
print(bao_cao(df, "thanh_pho"))

print("\nBáo cáo theo phương thức thanh toán:")
print(bao_cao(df, "thanh_toan"))

print("\nBáo cáo theo loại khách hàng:")
print(bao_cao(df, "loai_khach"))
```
